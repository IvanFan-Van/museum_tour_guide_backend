```python
import os
from supabase import create_client, Client
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv(), override=True)

supabase: Client = create_client(
    os.environ.get("SUPABASE_URL", ""), os.environ.get("SUPABASE_KEY", "")
)
```

```python
results = (
    supabase.table("sections")
    .select("section_idx", "title", "page_idx", "en_text")
    .execute()
)
```

```python
import qrcode
from pathlib import Path
from tqdm import tqdm

output_folder = Path("../output/qr_codes")
output_folder.mkdir(exist_ok=True)

for res in tqdm(results.data):
    # print(res)
    title = res["title"]
    idx = res["section_idx"]
    page_idx = res["page_idx"]
    text = res["en_text"]
    folder = Path(output_folder / f"{page_idx}_{title}")
    folder.mkdir(exist_ok=True)
    with open(folder / "text.md", "w") as f:
        f.write(text)

    url = f"https://museum.share.zrok.io?doc_id={idx}"
    print(url)
    img = qrcode.make(url)
    img.save(folder / "qr_code.png")
```

```python
import zipfile

# recursively iterate output_folder and put the complete output_folder into a zip file
with zipfile.ZipFile("../output/qr_codes.zip", "w") as zip_file:
    for folder in output_folder.iterdir():
        if folder.is_dir():
            for file in folder.iterdir():
                zip_file.write(file, arcname=f"{folder.name}/{file.name}")
```
